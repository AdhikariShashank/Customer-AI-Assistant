```python
from pathlib import Path

pdf_directory = Path("../data/pdfs")
print("PDF Directory: ", pdf_directory)

pdf_files = list(pdf_directory.glob("*.pdf"))
print("PDF Files: ", pdf_files)

```

```python
# %pip install Pymupdf
import fitz  # PyMuPDF

doc = fitz.open(pdf_files[0])  # Select the first PDF file

for current_page, page in enumerate(doc, start=1):
    text = page.get_text()
    print(f"Page {current_page}:\n{text}\n{'-'*40}\n")
```
